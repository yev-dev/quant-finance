## Yield Curves and Bootstrapping

This section covers yield-curve construction: short-rate instruments (deposits), money-market, OIS, and swap bootstrapping.
- Objective: build a zero/discount curve from market instruments and visualize the term structure.
- Starter code: a minimal illustrative bootstrap from deposit rates (next cell).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Simple bootstrap from short deposit rates (illustrative)
deposit_rates = {'1M': 0.005, '3M': 0.007, '6M': 0.01, '1Y': 0.015}
tenors_months = {'1M':1,'3M':3,'6M':6,'1Y':12}
def months_to_years(m):
    return m/12.0

rows = []
for tname, r in deposit_rates.items():
    T = months_to_years(tenors_months[tname])
    df = 1.0 / (1.0 + r * T)  # simple discount from deposit rate
    zero_cont = -np.log(df) / T
    rows.append({'tenor': tname, 'T': T, 'rate': r, 'discount': df, 'zero_cont': zero_cont})
df = pd.DataFrame(rows)
print(df)
plt.plot(df['T'], df['zero_cont'], marker='o')
plt.xlabel('Years')
plt.ylabel('Continuous zero rate')
plt.title('Simple bootstrapped zeros (illustrative)')
plt.grid(True)
plt.show()
```

## FX Forwards (Covered Interest Parity)

Covered interest parity (CIP) links spot and forward rates via interest rate differentials. For continuous compounding: \(F = S \exp((r_d - r_f)T)\).
- Objective: compute forwards from yields and visualize tenor-dependency.
- Starter code: simple forward formula (next cell).

```python
import numpy as np
# FX forward calculation using continuously compounded rates
spot = 1.1000  # spot FX rate (base/quote)
rd = 0.02      # domestic continuous rate
rf = 0.005     # foreign continuous rate
T = 0.5        # time in years
forward = spot * np.exp((rd - rf) * T)
print(f"Spot: {spot}, rd: {rd}, rf: {rf}, T: {T}\nForward: {forward:.6f}")
```

## Vol Surfaces

This section introduces implied volatility surfaces: how they are represented (expiry × strike), common parametrizations (SABR, SVI, local vol), and visualization techniques.
- Objective: visualize a surface, understand skew and term structure.
- Starter code: mock surface + plotting example in the following cell.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Mock implied vol surface: expiries (years) x strikes
expiries = np.array([0.1, 0.25, 0.5, 1.0, 2.0])
strikes = np.linspace(0.8, 1.2, 9)
E, S = np.meshgrid(expiries, strikes)
S0 = 1.0
base = 0.12
a = -0.03
b = 0.02
Vol = base + a * np.log(S / S0) + b * np.sqrt(E)

fig = plt.figure(figsize=(8, 5))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(E, S, Vol, cmap='viridis')
ax.set_xlabel('Expiry (y)')
ax.set_ylabel('Strike')
ax.set_zlabel('Vol')
plt.title('Mock Implied Vol Surface')
plt.tight_layout()
plt.show()
```
