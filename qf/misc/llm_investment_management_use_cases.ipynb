# 1. What is Alpha?

In investment management, **alpha** generally means investment return that cannot simply be explained by taking common market risks.

A simplified mental model is:

\[
\text{Portfolio Return} = \text{Systematic Return} + \text{Alpha}
\]

Suppose a portfolio returns 12% while a benchmark returns 8%.

It is tempting to say:

\[
\alpha = 12\% - 8\% = 4\%
\]

But this may be misleading. The portfolio may have had greater exposure to:

- equities in general,
- small companies,
- value stocks,
- momentum,
- technology,
- credit risk,
- emerging markets,
- illiquidity,
- leverage.

If those exposures explain the additional return, it is not necessarily manager-specific alpha.

A single-factor model can be written as:

\[
R_p - R_f = \alpha + \beta_m (R_m - R_f) + \epsilon
\]

A multi-factor model can be written as:

\[
R_p - R_f =
\alpha +
\beta_m MKT +
\beta_v VALUE +
\beta_s SIZE +
\beta_q QUALITY +
\beta_{mom} MOMENTUM +
\epsilon
\]

where alpha is the component not explained by the factors in the model.

## Why this matters for GenAI

An LLM should not merely produce convincing narratives. It should help the analyst answer:

- What does the market already know?
- What assumptions are reflected in price?
- Where might consensus be wrong?
- What evidence contradicts the current thesis?
- Have we seen this economic situation before?
- Is a seemingly new insight just a known factor exposure?
- Which new theme could create a differentiated investment opportunity?

A useful framing is:

> **GenAI supports the process that may generate alpha; it should not be confused with alpha itself.**

# 2. What is Factor Investing?

A **factor** is a measurable characteristic associated with systematic differences in risk or expected returns across securities.

Common equity factors include:

- **Value**
- **Momentum**
- **Quality**
- **Size**
- **Low volatility**
- **Profitability**
- **Investment**
- **Liquidity**

## Value

A value strategy systematically favours securities that are inexpensive relative to fundamentals.

Possible measures:

- P/E
- EV/EBITDA
- Price/Book
- Free-Cash-Flow Yield
- Earnings Yield

Example:

| Company | Price | EPS | P/E |
|---|---:|---:|---:|
| A | 100 | 10 | 10x |
| B | 100 | 4 | 25x |

A simple value screen may prefer Company A.

## Quality

Quality often refers to companies with:

- strong profitability,
- high ROIC,
- robust balance sheets,
- stable earnings,
- good cash conversion,
- disciplined capital allocation.

## Momentum

Momentum captures persistence in relative price performance.

One simplified signal is:

\[
Momentum = \frac{P_{t-1}}{P_{t-12}} - 1
\]

## Size

The size factor distinguishes smaller companies from larger companies.

## Factor investing versus fundamental investing

**Factor investing** asks:

> Which measurable characteristics have historically been rewarded across large groups of securities?

**Fundamental investing** asks:

> What is this particular business worth, and is the market mispricing it?

A fundamental portfolio can still have strong factor exposures.

For example, a contrarian fundamental manager may unintentionally carry:

- positive value exposure,
- negative momentum exposure,
- quality exposure.

That matters for performance attribution.

````python
import numpy as np
import pandas as pd

np.random.seed(42)
n = 120  # 10 years of monthly data

factors = pd.DataFrame({
    "Market": np.random.normal(0.006, 0.04, n),
    "Value": np.random.normal(0.002, 0.025, n),
    "Momentum": np.random.normal(0.003, 0.03, n),
    "Quality": np.random.normal(0.002, 0.02, n),
})

true_alpha = 0.002

portfolio = (
    true_alpha
    + 0.9 * factors["Market"]
    + 0.4 * factors["Value"]
    - 0.1 * factors["Momentum"]
    + 0.3 * factors["Quality"]
    + np.random.normal(0, 0.015, n)
)

df = factors.copy()
df["Portfolio"] = portfolio
df.head()
````

# 4. Company Fundamentals

**Company fundamentals** are the economic and financial characteristics that help determine the intrinsic value of a business.

## Revenue

Questions:

- Is revenue growing?
- Is growth driven by price or volume?
- Is growth organic or acquisition-driven?
- Which products and geographies drive growth?
- Is market share rising or falling?

## Gross Margin

\[
Gross\ Margin =
\frac{Revenue - Cost\ of\ Goods\ Sold}{Revenue}
\]

## Operating Margin

\[
Operating\ Margin =
\frac{Operating\ Profit}{Revenue}
\]

## Net Margin

\[
Net\ Margin =
\frac{Net\ Income}{Revenue}
\]

Margins help distinguish growth from profitable growth.

## Free Cash Flow

A simplified expression:

\[
FCF \approx Operating\ Cash\ Flow - Capital\ Expenditure
\]

Cash matters because it can be:

- reinvested,
- used to repay debt,
- paid as dividends,
- used for buybacks,
- used for acquisitions.

## Balance Sheet

Important questions:

- How much debt exists?
- What is the maturity profile?
- Can the company service interest?
- Does it have enough liquidity?
- How cyclical is its leverage?

Common metric:

\[
Net\ Debt / EBITDA
\]

## Return on Invested Capital

\[
ROIC = \frac{NOPAT}{Invested\ Capital}
\]

A company that sustainably earns a return above its cost of capital is creating economic value.

## Capital Allocation

Management can:

1. reinvest,
2. acquire,
3. repay debt,
4. pay dividends,
5. buy back shares.

Capital allocation can be one of the largest determinants of long-run shareholder value.

## Valuation

Common techniques:

- Discounted Cash Flow
- P/E
- EV/EBITDA
- Price/Book
- Free-Cash-Flow Yield
- Sum of the Parts
- Residual Income

The key question is not:

> "Is this a good company?"

It is:

> **"What expectations are embedded in the current price, and are they wrong?"**

# 5. Variant Perception

**Variant perception** means the investor's view of the future differs materially from the expectations embedded in the market price.

Example:

| Assumption | Market View | Investor View |
|---|---:|---:|
| Revenue growth | 2% | 6% |
| Operating margin | 10% | 14% |
| Capex intensity | High | Normalising |
| Competitive position | Weakening | Stable |

An opportunity may exist when:

1. the market prices one set of assumptions,
2. the investor holds a differentiated view,
3. the investor has strong evidence,
4. the difference is economically meaningful,
5. the investor is ultimately correct.

This is a natural area for LLMs because the market view is often scattered across:

- analyst reports,
- transcripts,
- news,
- filings,
- investor presentations,
- valuation assumptions,
- internal research.

# 6. Use Case — Thematic Investment Discovery

Thematic investment discovery is one of the most interesting non-obvious LLM use cases.

A **theme** is a structural change that can affect many companies across sectors and geographies.

Examples:

- AI infrastructure
- ageing populations
- electrification
- reshoring
- defence spending
- grid modernisation
- obesity drugs
- water scarcity
- semiconductor sovereignty
- automation
- energy transition

The challenge is that by the time a theme is obvious and widely discussed, much of the opportunity may already be reflected in prices.

The more interesting problem is:

> **Can an LLM detect weak signals that indicate an emerging structural theme before it becomes consensus?**

## Traditional thematic research

A human analyst might manually read:

- earnings calls,
- supplier commentary,
- regulation,
- patents,
- industry reports,
- capex announcements,
- procurement data,
- technical publications.

## Example

Suppose an LLM notices the following across unrelated companies:

- utilities discussing grid constraints,
- transformer manufacturers extending lead times,
- data-centre operators warning about power availability,
- copper suppliers increasing capex,
- governments accelerating grid investment.

Individually, each statement may appear unimportant.

Collectively they may indicate:

> **Power-grid bottlenecks are becoming a structural constraint on AI/data-centre growth.**

The investment opportunities may not be obvious AI companies.

They may be:

- transformers,
- switchgear,
- grid equipment,
- power management,
- copper,

- cooling infrastructure.

This is where an LLM can help connect information across conventional sector boundaries.

## 6.1 Theme Discovery vs Theme Storytelling

This distinction is important in an interview.

A weak GenAI application does:

> "Write me an investment theme about AI."

A stronger application does:

> **"Identify clusters of new evidence that have become more common, more economically significant, and more broadly confirmed across independent sources."**

The system should distinguish:

### Narrative popularity

How often is a theme mentioned?

from

### Fundamental significance

Is it causing changes in:

- revenue,
- capex,
- pricing,
- margins,
- market share,
- supply constraints,
- regulatory behaviour?

and from

### Investability

Which listed companies have economically meaningful exposure?

and from

### Valuation attractiveness

Is the opportunity already fully priced?

A theme can be real but still be a bad investment if valuation is excessive.

## 6.2 Theme Knowledge Graph

An LLM can help construct a graph such as:

```text
                   AI Data Centres

          ┌──────────────┼──────────────┐
          │              │              │
       Compute        Cooling         Power
          │              │              │
```

The most interesting investment may sit several steps away from the headline theme.

This is sometimes called identifying:

- second-order beneficiaries,
- picks-and-shovels providers,
- bottleneck owners,
- enabling infrastructure,
- hidden beneficiaries.

LLMs are useful because they can connect natural-language evidence to companies and economic relationships.

## 6.3 Thematic Discovery Scoring

A theme engine could score themes across several dimensions.

Example:

\[
ThemeScore =
w_1 Novelty +
w_2 EvidenceBreadth +
w_3 EvidenceAcceleration +
w_4 FundamentalImpact +
w_5 CrossSourceConfirmation
\]

Possible components:

### Novelty

Is the theme genuinely emerging, or already ubiquitous?

### Evidence breadth

How many independent companies or sources mention it?

### Evidence acceleration

Are mentions increasing?

### Fundamental impact

Is it changing capex, pricing, revenue, margins or strategic decisions?

### Cross-source confirmation

Is evidence coming from independent sources rather than one repeated narrative?

The output should not be:

> "Buy this theme."

It should be:

> **"This structural pattern is emerging. Here is the evidence, companies affected, confidence, contradictions, and potential investable exposures."**

````python
import pandas as pd

themes = pd.DataFrame({
    "theme": [
        "Grid modernisation",
        "AI software adoption",
        "Industrial reshoring",
        "Water infrastructure"
    ],
    "novelty": [0.8, 0.3, 0.6, 0.75],
    "evidence_breadth": [0.9, 1.0, 0.7, 0.65],
    "evidence_acceleration": [0.9, 0.7, 0.6, 0.8],
    "fundamental_impact": [0.95, 0.65, 0.75, 0.85],
    "cross_source_confirmation": [0.9, 0.9, 0.7, 0.8]
})

weights = {
    "novelty": 0.20,
    "evidence_breadth": 0.20,
    "evidence_acceleration": 0.20,
    "fundamental_impact": 0.25,
    "cross_source_confirmation": 0.15
}

themes["theme_score"] = sum(
    themes[col] * weight for col, weight in weights.items()
)

themes.sort_values("theme_score", ascending=False)
````

## 6.4 Backtesting Thematic Discovery

A major risk is **hindsight bias**.

It is easy to look back and say:

> "AI infrastructure was obviously a major theme."

The real test is:

> **Could the system have identified the theme using only information that existed at the time?**

A point-in-time thematic backtest should:

1. Choose historical dates.
2. Freeze the information universe at each date.
3. Run the theme-discovery process.
4. Record themes without looking at future outcomes.
5. Map the themes to investable companies.
6. Measure subsequent outcomes.
7. Compare against benchmarks and factor exposures.

Possible metrics:

- theme emergence lead time,
- precision of identified beneficiaries,
- subsequent revenue acceleration,
- earnings revision direction,
- stock return,
- factor-adjusted return,
- analyst usefulness.

The research question is not just:

> "Did the theme outperform?"

It is also:

> "Did the system discover a useful structural change earlier than a conventional search or analyst workflow?"

# 7. Use Case — Investment Analogue Engine

A normal search engine retrieves documents with similar words.

An investment analogue engine tries to retrieve **similar economic situations**.

Example current situation:

- margins have collapsed,
- market sentiment is poor,
- demand may be cyclical,
- balance sheet is strong,
- management continues to invest,
- analyst believes the market is treating temporary weakness as structural decline.

The system should retrieve historical cases based on economic structure rather than company name.

```text
Current Company
      │
      ▼
Extract Investment Situation
      │
      ├── margin compression
      ├── negative sentiment
      ├── strong balance sheet
      ├── reinvestment
      └── temporary-vs-structural debate
      │
      ▼
Search Historical Research
      │
      ▼
Retrieve Analogues
      │
      ▼
Compare Outcomes and Lessons
```

Example output:

> "Five historical situations were similar. In three, margins normalised. In two, the apparent cyclical weakness became structural. The strongest discriminator was persistent market-share loss."

This turns proprietary historical research into **institutional memory**.

# 8. Use Case — Thesis Challenger / Red-Team Agent

Instead of asking the model to support an analyst, explicitly ask it to challenge the thesis.

Example thesis:

> Company X is undervalued because the market overestimates the permanence of its margin decline.

Decompose it:

```text
Investment Thesis
    │
    ├── Demand normalises
    ├── Market share remains stable
    ├── Input costs fall
    ├── Management maintains pricing
    └── Balance sheet survives downturn
```

The agent searches for evidence that falsifies each assumption.

Example:

```text
Assumption:
Market share remains stable

Evidence FOR
- historical retention
- stable customer base

Evidence AGAINST
- competitor capacity expansion
- distributor comments
- weaker renewals

Assessment:
THESIS COMPONENT WEAKENING
```

This is substantially more useful than generic summarisation.

# 17. RAG for Investment Research

## What is RAG?

**Retrieval-Augmented Generation** combines:

1. retrieval of relevant information,
2. LLM reasoning over that retrieved evidence.

Without RAG:

```text
Question
   ↓
LLM
   ↓
Answer based mainly on model memory
```

With RAG:

```text
Question
   ↓
Retrieval
   ↓
Relevant Evidence
   ↓
Context Assembly
   ↓
LLM
   ↓
Grounded Answer + Citations
```

RAG is especially important in investment research because:

- information changes constantly,
- internal research is private,
- citations matter,
- historical versions matter,
- hallucination risk is unacceptable.

A robust investment RAG system normally combines all three.

# 21. Backtesting

Backtesting asks:

> **How would a strategy or signal have behaved historically if it had been available at the time?**

For a conventional quantitative strategy:

1. calculate the signal,
2. rank securities,
3. form a portfolio,
4. advance through time,
5. calculate returns.

For an LLM system, backtesting is more subtle.

We may want to test:

- an LLM-derived sentiment signal,
- management credibility,
- thesis-risk score,
- thematic exposure,
- cyclical-vs-structural classification,
- analogue similarity,
- contradiction intensity.

# 22. Point-in-Time Data and Look-Ahead Bias

This is essential.

Suppose we test a 2020 investment.

The model must only see information available in 2020.

It must **not** see:

- later filings,
- revised databases,
- future transcripts,
- future analyst notes,
- future outcomes.

Otherwise we create **look-ahead bias**.

Correct design:

```text
Historical Date T
       │
       ▼
Freeze Information Set
       │
       ▼
Run LLM / RAG
       │
       ▼
Generate Signal
       │
       ▼
Advance to T + 1
       │
       ▼
Measure Outcome
```

In investment AI, temporal integrity is one of the most important evaluation controls.

````python
import pandas as pd
import numpy as np

np.random.seed(7)

dates = pd.date_range("2018-01-31", periods=72, freq="ME")
assets = ["A", "B", "C", "D", "E", "F"]

rows = []
for date in dates:
    for asset in assets:
        signal = np.random.normal()
        future_return = 0.01 * signal + np.random.normal(0, 0.05)
        rows.append([date, asset, signal, future_return])

bt = pd.DataFrame(rows, columns=[
    "date", "asset", "llm_signal", "next_month_return"
])

bt["rank"] = bt.groupby("date")["llm_signal"].rank(pct=True)

bt["position"] = np.where(
    bt["rank"] >= 0.8, 1,
    np.where(bt["rank"] <= 0.2, -1, 0)
)

monthly = bt.groupby("date").apply(
    lambda x: (x["position"] * x["next_month_return"]).mean(),
    include_groups=False
)

print("Toy average monthly strategy return:", monthly.mean())
print("Toy annualised approximation:", monthly.mean() * 12)
````

# 23. Backtesting an LLM Research Signal

Imagine we create a **management credibility score**.

At each historical quarter:

1. retrieve only documents available at that time,
2. extract prior management promises,
3. compare previous promises with realised outcomes,
4. calculate credibility,
5. generate a score,
6. relate the score to future returns or fundamental outcomes.

But price return should not be the only metric.

We can also test whether credibility predicts:

- forecast misses,
- earnings revisions,
- margin surprises,
- capital-allocation mistakes,
- leverage deterioration.

That often gives a better causal interpretation than jumping directly to stock returns.

# 24. Factor-Neutral Backtesting

Suppose an LLM score predicts returns.

We still need to ask:

> Is the LLM simply rediscovering value or momentum?

A conceptual regression:

\[
R_{t+1} =
\alpha +
\beta_1 LLMScore_t +
\beta_2 Value_t +
\beta_3 Momentum_t +
\beta_4 Quality_t +
\epsilon
\]

If the coefficient on `LLMScore` remains informative after factor controls, the signal is more interesting.

Another approach is portfolio neutralisation:

- rank companies by LLM score,
- neutralise sector,
- neutralise country,
- neutralise value,
- neutralise market beta,
- compare residual performance.

# Fine-Tuning LLM Weights for Beginners (And How It Compares to RAG)

Before the no-retraining approach, it helps to understand the alternative: **fine-tuning model weights**.

## 1) What Fine-Tuning Means

Fine-tuning updates model parameters using your labeled examples so the model internalizes a domain-specific behavior.

- You prepare instruction-response pairs or preference data.
- You run additional training steps on a base model.
- The output is a new model checkpoint (or adapter weights) specialized for your task.

## 2) Common Fine-Tuning Libraries and Tools

### Python Training Stack
- `transformers` (Hugging Face): model loading, tokenization, training APIs.
- `datasets`: dataset loading and preprocessing.
- `trl`: supervised fine-tuning and preference optimization workflows.
- `peft`: parameter-efficient methods like LoRA/QLoRA.
- `accelerate`: distributed and mixed-precision training orchestration.
- `bitsandbytes`: quantization and memory-efficient training.

### Managed / Platform Options
- OpenAI fine-tuning APIs (for supported models).
- Cloud managed tuning services (AWS/GCP/Azure equivalents).
- Inference frameworks after tuning: `vLLM`, `TGI`, or vendor-hosted endpoints.

## 3) Beginner Architecture for Fine-Tuning

1. Define target task: for example earnings-call summarization style, or risk memo drafting format.
2. Build high-quality training set: clean, deduplicated, policy-compliant examples.
3. Choose method: full fine-tune vs LoRA/QLoRA adapter tuning.
4. Train on validation split and track quality metrics.
5. Evaluate against baseline model and against a RAG-only baseline.
6. Deploy tuned model endpoint and monitor drift/cost/safety.

## 4) Pros and Cons: Fine-Tuning vs No-Retraining (RAG + Tools)

### Fine-Tuning Weights
**Pros**
- Better consistency in tone/format for repeated tasks.
- Can improve behavior on narrow tasks with strong labeled data.
- Lower prompt complexity once behavior is learned.

**Cons**
- Requires curated training data and ML training workflow.
- Risk of memorizing stale patterns if data changes quickly.
- Harder to keep factual content fresh without retrieval.
- More ops burden: versioning, evals, rollback, deployment lifecycle.

### No-Retraining (RAG + Prompt + Tools)
**Pros**
- Fast iteration and lower initial cost.
- Easy to keep answers fresh with latest documents/data.
- Better traceability via citations and source links.
- Strong fit when financial facts change daily.

**Cons**
- Requires strong retrieval/index engineering.
- Output style may vary more than tuned models.
- Needs robust context-window management and ranking.

## 5) Practical Decision Rule for Investment Research

Use **RAG-first** when:
- data freshness and auditability are top priorities,
- knowledge updates frequently (markets, filings, transcripts),
- you need citations and explainability for investment committees.

Add **fine-tuning** later when:
- you already have reliable retrieval and tool pipelines,
- you want consistent writing style/workflow behavior,
- you have enough high-quality labeled examples to justify training.

## 6) Recommended Hybrid Strategy

For most GenAI investment research platforms:
1. Start with RAG + tools + guardrails.
2. Measure failure modes (style inconsistency, reasoning format gaps).
3. Fine-tune small adapters (LoRA) only for the gaps that persist.
4. Keep RAG in place for factual freshness and provenance.

This gives you both: **fresh evidence** and **stable task behavior**.

# Custom LLM for Investment Research Without Retraining

> Goal: build a domain-specialized GenAI assistant for investment research **without** training a new foundation model or fine-tuning weights.

## 1) Options for Customization Without Retraining

> Think in layers: prompt layer, retrieval layer, tools layer, control layer.

1. **Prompt/System Design**
   - Encode role, tone, and output schema (memo, risk brief, IC note).
   - Force evidence-first answers and explicit uncertainty statements.
   - Add reusable templates for comparable-company analysis, earnings summaries, and portfolio attribution.

2. **RAG (Retrieval-Augmented Generation)**
   - Keep the model generic; customize with **your data** at inference time.
   - Retrieve from research notes, earnings transcripts, filings, models, CRM notes, and internal investment theses.
   - Ground responses with citations to source passages and timestamps.

3. **Tool-Augmented LLM (Function Calling / Agents)**
   - Let the model call deterministic tools: SQL, valuation calculators, factor exposures, scenario engine, backtest utilities.
   - Keep critical math and data transforms outside the LLM for reliability.

4. **Model Routing / Ensemble Orchestration**
   - Route simple requests to cheaper models and complex research tasks to stronger models.
   - Add policy routing for confidential vs public data workloads.

5. **Guardrails + Evaluation Layer**
   - Add policies for data leakage, prohibited advice phrasing, unsupported claims, and stale-data warnings.
   - Evaluate factuality, citation coverage, answer completeness, and latency/cost.

## 2) Custom AI Context Design (What Makes It "Your" Model)

Your model becomes custom through **context engineering**:

- **Domain ontology**: issuer, security, strategy, factor, region, vintage, deal stage.
- **Canonical identifiers**: map ticker/ISIN/CUSIP/internal IDs to one entity graph.
- **Data contracts**: freshness SLAs, nullable fields, reconciliation rules, source precedence.
- **User context**: team role (PM, analyst, risk), coverage universe, watchlist, mandate constraints.
- **Task context**: decision type (initiate/trim/exit), horizon, benchmark, risk budget.
- **Memory context**: prior meeting notes, previous model outputs, open questions, unresolved risks.

## 3) Reference Architecture for a GenAI Investment Research Platform

> No retraining needed: the intelligence comes from retrieval, tools, and governance.

- **Ingestion**: market data, alt/private data, filings, transcripts, broker research, internal notes.
- **Normalization**: entity resolution, schema harmonization, quality checks, PII/classification tags.
- **Storage**:
  - vector index (semantic retrieval)
  - keyword index (exact search)
  - relational warehouse/lakehouse (metrics, history, joins)
  - knowledge graph (entity relationships)
- **Retrieval orchestration**: hybrid search + reranker + context pack builder.
- **LLM orchestration**: planner, prompt builder, tool-calling policy, response synthesis.
- **Guardrails**: policy checks pre- and post-generation.
- **Evaluation/observability**: trace logs, answer grading, cost/latency dashboards, drift/staleness alerts.

## 4) How RAG Is Linked End-to-End

1. User asks: "Should we increase exposure to X ahead of earnings?"
2. Query parser extracts entities, timeframe, intent, risk focus.
3. Retriever fetches top evidence from multiple stores (hybrid retrieval).
4. Reranker promotes the most relevant and freshest documents.
5. Context packer builds a bounded prompt with citations and metadata.
6. LLM drafts answer and calls tools for exact metrics where needed.
7. Guardrails verify claims, missing citations, and policy constraints.
8. Final response returns recommendation framing + evidence + uncertainty.

## 5) Implementation Blueprint (Practical Sequence)

- **Phase 1: Foundation**
  - Build document pipelines and hybrid retrieval.
  - Add strict citation format and "insufficient evidence" behavior.
- **Phase 2: Analyst Copilot**
  - Add tool-calling for SQL/factor/valuation utilities.
  - Add templates for investment memos and risk summaries.
- **Phase 3: Production Controls**
  - Introduce offline eval sets and online feedback loops.
  - Add role-based access and policy gating.
- **Phase 4: Portfolio Workflow Integration**
  - Integrate watchlists, rebalance triggers, and PM dashboard actions.
  - Add audit trails for every generated recommendation.
