```python
import pandas as pd

df = pd.DataFrame({
    "price": [100, 102, 101, 105, 110, 108, 112, 115, 117, 120, 125]
})

df["return_1d"] = df["price"].pct_change(1)
df["return_10d"] = df["price"].pct_change(10)

print(df)
```

```python
df["abs_return_1d"] = df["price"].diff(1)
df["abs_return_10d"] = df["price"].diff(10)
```

```python
df["price_lag_1d"] = df["price"].shift(1)
df["price_lag_10d"] = df["price"].shift(10)
```
