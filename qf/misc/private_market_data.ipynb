# Private Markets Terminology — Beginner Guide

This notebook explains the main private-market terms used in technical interviews, data engineering, analytics, and AI/ML use cases.

The goal is not to become a private-equity specialist. You only need enough domain knowledge to understand what the data represents, how important metrics are calculated, why data quality matters, why anomaly detection is useful, and why time/versioning concepts are important in financial systems.


## NAV — Net Asset Value

**NAV** stands for **Net Asset Value**.

It represents the estimated value of a fund's remaining assets after subtracting liabilities.

\[
NAV = Assets - Liabilities
\]

### Simple example

Suppose a private-equity fund owns:

- Company A: £40m
- Company B: £35m
- Company C: £25m

Total assets = £100m.

If the fund has £10m of liabilities:

\[
NAV = £100m - £10m = £90m
\]

So the fund's NAV is **£90 million**.

### Why NAV matters in private markets

Publicly traded companies usually have observable market prices. Private companies usually do not. Therefore, private-market investments are valued periodically rather than continuously.

```text
31 Mar 2026   NAV = £100m
30 Jun 2026   NAV = £108m
30 Sep 2026   NAV = £115m
```


````python
assets = {"Company A": 40_000_000, "Company B": 35_000_000, "Company C": 25_000_000}
liabilities = 10_000_000
total_assets = sum(assets.values())
nav = total_assets - liabilities
print(f"Total assets: £{total_assets/1_000_000:.0f}m")
print(f"Liabilities: £{liabilities/1_000_000:.0f}m")
print(f"NAV: £{nav/1_000_000:.0f}m")
````

## GP — General Partner

The **GP** is the organisation that manages the private-market fund.

Typical responsibilities include raising the fund, selecting investments, managing portfolio companies, valuing investments, selling assets, and reporting fund performance.


## LP — Limited Partner

The **LP** is an investor in the fund.

Typical LPs include pension funds, insurance companies, sovereign wealth funds, endowments, family offices, and asset managers.


## Commitment

A **commitment** is the amount of money an LP agrees to make available to a fund.

```text
LP commitment = £100m
```

The investor does not necessarily transfer all of it immediately.


## Capital Call

A **capital call** is money requested from investors by the GP.

```text
Commitment:      £100m
Capital called:   £30m
Remaining:        £70m
```

Capital calls may also be described as contributions, drawdowns, or paid-in capital depending on context.


## Paid-In Capital

**Paid-In Capital** is the amount investors have actually contributed so far.

```text
Original commitment: £100m
Capital called:       £60m
Paid-In Capital:      £60m
```


## Distribution

A **distribution** is money returned from the fund to investors.

It may result from an asset sale, investment income, debt repayment, or partial realisation.


## Realised vs Unrealised Value

**Realised value** has already been converted into cash and returned or made available to investors.

**Unrealised value** is still held inside the fund and contributes to NAV.


## DPI — Distributed to Paid-In

\[
DPI = \frac{Distributions}{Paid\text{-}In\ Capital}
\]

Example:

```text
Paid-In Capital = £100m
Distributions   = £40m
DPI             = 0.4x
```

This means investors have received cash equal to 40% of the amount contributed.


## RVPI — Residual Value to Paid-In

\[
RVPI = \frac{NAV}{Paid\text{-}In\ Capital}
\]

Example:

```text
NAV             = £90m
Paid-In Capital = £100m
RVPI            = 0.9x
```


## TVPI — Total Value to Paid-In

\[
TVPI = \frac{Distributions + NAV}{Paid\text{-}In\ Capital}
\]

Example:

```text
Paid-In Capital = £100m
Distributions   = £40m
NAV             = £90m
TVPI            = 1.3x
```

Important relationship:

\[
TVPI = DPI + RVPI
\]


````python
paid_in = 100_000_000
distributions = 40_000_000
nav = 90_000_000

dpi = distributions / paid_in
rvpi = nav / paid_in
tvpi = (distributions + nav) / paid_in

print(f"DPI:  {dpi:.2f}x")
print(f"RVPI: {rvpi:.2f}x")
print(f"TVPI: {tvpi:.2f}x")
print(f"Check: DPI + RVPI = {(dpi + rvpi):.2f}x")
````

## IRR — Internal Rate of Return

**IRR** measures investment performance while taking the timing of cashflows into account.

£10m returned after 1 year is financially different from £10m returned after 10 years.

For interviews, remember:

> TVPI measures total value relative to invested capital, while IRR also considers how quickly that value was created.


## Vintage Year

The **vintage year** usually refers to the year in which a fund started investing or first called capital.

Funds are often benchmarked against others with a similar vintage.


## Fund Strategy

Common private-market strategies include:

- Private Equity
- Venture Capital
- Growth Equity
- Private Debt
- Infrastructure
- Real Estate
- Secondaries


## Dry Powder

**Dry powder** is committed capital that has not yet been invested.

\[
Dry\ Powder = Committed\ Capital - Capital\ Deployed
\]


## J-Curve

Private-market funds often show weak or negative returns in early years before improving later.

This pattern is called the **J-curve**.

Early weakness may reflect fees, transaction costs, and investments that have not yet appreciated.


## Benchmark

A **benchmark** allows a fund to be compared with similar funds.

Peer groups may use strategy, vintage year, geography, and fund size.

Benchmarks can calculate median IRR, median TVPI, quartiles, and percentiles.


## Quartile

A **quartile** divides a peer group into four groups.

```text
Top Quartile       Best 25%
Second Quartile    25–50%
Third Quartile     50–75%
Bottom Quartile    Worst 25%
```


## Effective Date vs Received Date

**Effective date** is when the financial information applies.

**Received date** is when your system actually obtained it.

Example:

```text
NAV effective date = 30 June 2026
Received date      = 18 August 2026
```

This distinction matters for historical reconstruction, auditability, ML leakage, and benchmark reproducibility.


## Observation

An **observation** is a value received from a particular source.

```text
Fund:          ABC Fund
Metric:        NAV
Value:         £105m
Effective:     30 Jun 2026
Received:      15 Jul 2026
Source:        GP Report
```

A platform may preserve several observations for the same metric instead of overwriting them.


## Canonical Value

A **canonical value** is the trusted value selected by the platform after validation and reconciliation.

A good system should retain which observation was selected, why it was selected, and which methodology or version was used.


## Derived Data

**Derived data** is data calculated from other data rather than received directly from a source.

Input data might include:

```text
Paid-In Capital
Distributions
NAV
Cashflow Dates
```

Derived outputs might include:

```text
DPI
RVPI
TVPI
IRR
Quartile
Benchmark
Anomaly Score
```

Conceptually:

```text
Raw / Source Data
       ↓
Validation
       ↓
Canonical Data
       ↓
Calculations
       ↓
Derived Data
```


## Data Lineage

**Data lineage** describes where a value came from and how it was produced.

```text
TVPI = 1.3x
   ↓
Distribution = £40m
NAV          = £90m
Paid-In      = £100m
   ↓
Source observations
   ↓
Original GP / LP reports
```

Lineage is important for debugging, audit, controls, explainability, and reproducibility.


## Data Quality

Typical data-quality dimensions include:

- Completeness
- Accuracy
- Consistency
- Validity
- Uniqueness
- Freshness
- Timeliness

Examples:

```text
Missing fund ID              -> completeness problem
Negative NAV                 -> validity problem
Duplicate cashflow           -> uniqueness problem
Different values by source   -> consistency problem
Data not updated for 2 years -> freshness problem
```


## Anomaly

An **anomaly** is an observation that looks unusual compared with expected behaviour.

```text
Previous NAV = £100m
New NAV      = £105m
```

may look normal, whereas:

```text
Previous NAV = £100m
New NAV      = £600m
```

may deserve investigation.

An anomaly is **not automatically an error**. It may represent a genuine financial event.

A good system often detects the anomaly, assigns a score, provides an explanation, and flags it for review.
